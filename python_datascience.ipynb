### Estruturas Condicionais

#### Conceitos:

- `if:` executa um bloco de instruções caso a condição seja verdadeira;

- `else:` executa um bloco de instruções caso a condição seja falsa;

- `elif:` usado em casos de condicionais em cascata/cadeia;

- `operadores lógicos:` servem para analisar mais de uma condição no mesmo if
    - `and`: todas as condições precisam ser verdadeiras;
    - `or`: pelo menos uma condição precisa ser verdadeira;
    - `not`: negação de uma condição;
    - `in`: buscar dentro;

- Sintaxe:
```
if <condicao>:
    <instruções>
elif <condicao alternativa>:
    <instruções>
else:
    <instruções>
```

##### Exercício 4: Calculando o IMC

````python
print('Bem vindo ao Cálculo do IMC')

peso = float(input('Digite o seu peso (kg): '))
altura = float(input('Digite o seu altura (m): '))

imc = peso / (altura**2)

print(f'Seu IMC é: {imc:.2f}')
if imc < 18.5:
    print('Você está abaixo do peso!')
elif 18.5 <= imc < 25:
    print('Você está no peso normal.')
elif imc >= 25:
    print('Você está acima do peso!')
````

##### Exercício 5: Controlando o Orçamento Mensal

````python
print('Bem vindo ao Orçamento Mensal')

despesas = float(input('Digite o total de despesas do mês (R$): '))

limite = 3000
saldo = limite - despesas
excesso = despesas - limite

if despesas > limite:
    print(f'Você gastou R${despesas:.2f} e passou R${excesso:.2f} do limite')
elif despesas == limite:
    print('Você está exatamente no limite de R$3000,00.')
elif despesas < limite:
    print(f'Você gastou R${despesas:.2f} e ainda têm R${saldo:.2f} de limite.')
````

##### Exercício 10: Aprovando Empréstimos

````python
print('Bem vindo ao Emprestimo')

renda_mensal = float(input('Digite o valor da sua renda mensal (R$): '))
parcela = float(input('Digite o valor da parcela desejada (R$): '))

parcela_30 = 0.30 * renda_mensal

if renda_mensal >= 2000 and parcela <= parcela_30:
    print(f'Renda: {renda_mensal:.2f}')
    print(f'Parcela Desejada: {parcela:.2f}')
    print(f'30% da Parcela: {parcela_30:.2f}')
    print('Empréstimo aprovado!')
elif renda_mensal < 2000:
    print('Empréstimo negado: renda menor que R$2000,00.')
elif parcela > parcela_30:
    print('Empréstimo negado: parcela acima de 30% da renda.')
````

### Estruturas de Repetição

#### Conceitos:

- `while:`

    - repete o bloco desde que a condição seja verdadeira;

    - sempre verifica a condição primeiro;

    - precisa de contador e incrementação (`+= ou -=`)

```
contador =

while <condicao>:
    <instruções>
    contador ++ ou -- ou += ou -=
```



- `for:`

    - funciona sob iteração de um conjunto de elementos;

    - `range (inicio, fim, passo)`: o fim é excluso;

    - comandos de controle:
        - `continue`: salta para próxima iteração;
        - `break`: interrompe a execução do laço (loop);

```
for <elemento> in <conjunto>:
    <instruções>
```

#### Dicionários

- Dicionários:

    - são compostos por pares do tipo `{chave : valor }`;

    - delimitado por {  };

    - cada elemento do dicionário é um par;

    - o acesso do valor da chave é feito pelo nome da chave: `nome_dicionario['nome_chave'] => valor correspondente`

    - adicionando nova chave e valor a um dicionário já existente: `nome_dicionario['nome_nova_chave'] = valor`

    - chaves e valores podem ser de qualquer tipo;

````python
dicio = {'chave1':1, 'chave2':2}
print(dicio)

cadastro = {'matricula': 2000168933, 'dia de cadastro': 25, 'mês de cadastro': 10, 'turma': '2E'}
print(cadastro)
print(cadastro['matricula'])
print(cadastro['turma'])
````

````python
cadastro['turma'] = '2G'
cadastro
````

````python
cadastro['modalidade'] = 'EAD'
cadastro
````

#### Manipulação de Dicionários:

- Métodos em Dicionários:

    - `pop():` remove um item (especificado pela chave) e o retorna;

    - `items():` retorna todos os pares do dicionario;

    - `keys():` retorna apenas as chaves;

    - `values():` retorna apenas os valores;

    - consultar usando o laço FOR:
        - `for chaves in dicionario.keys()`
        - `for valores in dicionario.values()`

````python
cadastro.pop('turma')
print(cadastro)
````

````python
cadastro.items()
````

````python
cadastro.keys()
````

````python
cadastro.values()
````

````python
for chaves in cadastro.keys():
  print(chaves)
````

````python
for chaves in cadastro.keys():
  print(chaves)
````

````python
for chave, valor in cadastro.items():
  print(chave, valor)
````

````python
loja = {'nomes': ['televisão', 'celular', 'notebook', 'geladeira', 'fogão'],
        'precos': [2000, 1500, 3500, 4000, 1500]}

for chave, elementos in loja.items():
  print(f'Chave: {chave}\nElementos:')
  for dado in elementos:
    print(dado)
````
